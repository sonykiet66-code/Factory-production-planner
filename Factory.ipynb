```python
# ============================================================
# 🏭 FACTORY PRODUCTION PLANNER
# Mathematics: Linear Systems, Rank & Consistency
# Beginner-friendly Google Colab project
# ============================================================

import numpy as np
import matplotlib.pyplot as plt

print("=" * 55)
print("🏭 FACTORY PRODUCTION PLANNER")
print("=" * 55)

# ------------------------------------------------------------
# STEP 1: ENTER FACTORY DATA
# ------------------------------------------------------------
# A = machine-time coefficients
# B = required machine hours
#
# You can change these numbers to test another case.

A = np.array([
    [2, 1, 1],   # Machine 1
    [1, 2, 1],   # Machine 2
    [1, 1, 2]    # Machine 3
], dtype=float)

B = np.array([100, 80, 90], dtype=float)

print("\n📐 STEP 1: MATRIX FORM")
print("-" * 55)

print("Machine-time matrix A:")
print(A)

print("\nRequired machine hours B:")
print(B)

print("\nMathematical model:")
print("AX = B")


# ------------------------------------------------------------
# STEP 2: CREATE AUGMENTED MATRIX
# ------------------------------------------------------------

augmented = np.column_stack((A, B))

print("\n📊 STEP 2: AUGMENTED MATRIX [A|B]")
print("-" * 55)
print(augmented)


# ------------------------------------------------------------
# STEP 3: CALCULATE RANK
# ------------------------------------------------------------

rank_A = np.linalg.matrix_rank(A)
rank_augmented = np.linalg.matrix_rank(augmented)

print("\n🔢 STEP 3: RANK CALCULATION")
print("-" * 55)

print("Rank(A)      =", rank_A)
print("Rank([A|B])  =", rank_augmented)


# ------------------------------------------------------------
# STEP 4: CHECK CONSISTENCY
# ------------------------------------------------------------

print("\n✅ STEP 4: CONSISTENCY TEST")
print("-" * 55)

number_of_variables = A.shape[1]

if rank_A != rank_augmented:

    print("❌ NO SOLUTION")
    print("The system is INCONSISTENT.")
    print("Rank(A) ≠ Rank([A|B])")

    X = None

elif rank_A == number_of_variables:

    print("✅ UNIQUE SOLUTION")
    print("The system is CONSISTENT.")
    print("Rank(A) = Rank([A|B]) = Number of variables")

    X = np.linalg.solve(A, B)

else:

    print("⚠️ INFINITELY MANY SOLUTIONS")
    print("The system is CONSISTENT but does not have a unique solution.")

    X = None


# ------------------------------------------------------------
# STEP 5: CALCULATE PRODUCTION PLAN
# ------------------------------------------------------------

if X is not None:

    print("\n🏭 STEP 5: PRODUCTION PLAN")
    print("-" * 55)

    print(f"Product A = {X[0]:.2f} units")
    print(f"Product B = {X[1]:.2f} units")
    print(f"Product C = {X[2]:.2f} units")


# ------------------------------------------------------------
# STEP 6: VERIFY THE MATHEMATICAL ANSWER
# ------------------------------------------------------------

if X is not None:

    calculated_hours = A @ X

    print("\n🔍 STEP 6: VALIDATION")
    print("-" * 55)

    print("Required machine hours:  ", B)
    print("Calculated machine hours:", np.round(calculated_hours, 2))

    if np.allclose(calculated_hours, B):
        print("\n✅ VALIDATION PASSED")
        print("AX = B is satisfied.")
    else:
        print("\n❌ VALIDATION FAILED")


# ------------------------------------------------------------
# STEP 7: VISUALISE PRODUCTION QUANTITIES
# ------------------------------------------------------------

if X is not None:

    print("\n📈 STEP 7: PRODUCTION VISUALISATION")
    print("-" * 55)

    products = ["Product A", "Product B", "Product C"]

    plt.figure(figsize=(8, 5))
    plt.bar(products, X)

    plt.xlabel("Products")
    plt.ylabel("Production Quantity")
    plt.title("🏭 Factory Production Plan")

    # Display values above bars
    for i, value in enumerate(X):
        plt.text(i, value + 0.5, f"{value:.2f}",
                 ha="center")

    plt.grid(axis="y", alpha=0.3)
    plt.show()


# ------------------------------------------------------------
# STEP 8: FINAL PROJECT SUMMARY
# ------------------------------------------------------------

print("\n" + "=" * 55)
print("📋 PROJECT SUMMARY")
print("=" * 55)

print("""
FACTORY DATA
     ↓
MACHINE-TIME CONSTRAINTS
     ↓
LINEAR EQUATIONS
     ↓
MATRIX FORM AX = B
     ↓
RANK(A) & RANK([A|B])
     ↓
CONSISTENCY TEST
     ↓
PRODUCTION PLAN
     ↓
📊 VISUALISATION
""")

if rank_A == rank_augmented == number_of_variables:
    print("🎯 Final Result: UNIQUE & CONSISTENT PRODUCTION PLAN")
elif rank_A == rank_augmented:
    print("🎯 Final Result: CONSISTENT SYSTEM")
else:
    print("🎯 Final Result: INCONSISTENT SYSTEM")

print("=" * 55)
print("✅ Factory Production Planner completed!")
print("=" * 55)
```
